```python
list_data = [
    ('20/04/2026',305833,43.0,295.99,'Posto Bouganville'),
    ('09/05/2026',306148,38.3,260.62,'Posto Santo Afonso'),
    ('24/05/2026',306518,47.21,306.41,'Auto Posto Mandala'),
    ('08/06/2026',306868,38.29,259.72,'Posto de Gasolina BRA'),
    ('04/07/2026',307136,36.74,245.83,'Posto Santo Afonso'),
    ('24/07/2026',307474,37.68,255.85,'Posto Santo Afonso'),
    ('16/08/2026',307831,41.28,276.14,'Posto Santo Afonso'),
]

def calculate_fuel_economy(data):
    # date, odometer, liters, cost, station = data
    # fuel_economy = odometer / liters
    # return (date, fuel_economy)
    list_odometer = []
    list_liters = []
    list_index = []

    for data in list_data:
        i = list_data.index(data)
        if i > 0:
            previous_odometer = list_data[i - 1][1]
            current_odometer = data[1]
            odometer_difference = current_odometer - previous_odometer
            fuel_economy = odometer_difference / data[2]
            print(f"Date: {data[0]}, Fuel Economy: {fuel_economy:.2f} km/l no posto {data[4]}")
        list_odometer.append(data[1])
        list_liters.append(data[2])
        list_index.append(i)

    return list_odometer, list_liters, list_index


print(calculate_fuel_economy(list_data))

list_data[1][2]

              
```
